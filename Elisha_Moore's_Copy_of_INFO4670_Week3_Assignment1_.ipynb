### Q1 · One quantitative variable
**How many hours per week do Northgate students work?** Describe what's typical, how spread out it is, and whether anyone stands apart.
→ *histogram* (add a *box plot* if it helps). Column: `work_hours_per_week` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.histplot(data=students, x="work_hours_per_week", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("Work hours for Northgate students")
#   ax.set_xlabel("Time")
#   ax.set_ylabel("Number Of Students")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots()
sns.histplot(data=students, x="work_hours_per_week", color=NAVY, ax=ax) # which chart? which column(s)?
ax.set_title("Work hours for Northgate students")
ax.set_xlabel("Time")
ax.set_ylabel("Number Of Students")
plt.show()
```

### Q2 · One categorical variable
**How does enrollment compare across majors?**
→ *sorted bar chart*. Column: `major` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.countplot(data=students, x="Major", color=NAVY, ax=ax)   # bar chart which column(s)?
#   ax.set_title("Number of Students by Major")
#   ax.set_xlabel("Number of Students")
#   ax.set_ylabel("Major")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots()
sns.countplot(data=students, x="major", color=NAVY, ax=ax)
ax.set_title("Number of Students by Major")
ax.set_xlabel("Major")
ax.set_ylabel("Number of Students")
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for readability
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

**STAR story — Q2** *(write in your own words)*

- **Track —** The question that brought us to this chart is enrollment comparisons sdorted by each major.
- **Articulate —** The chart is specifying how data science, learning technologies, cybersecurity, information technology, and business analytics differentiate in numbers from different scales. For example, cybersecurity has over 500 students enrolled in that major.
- **Respond —** What should be examined is the repetitiveness of each major and being synchronized. While also looking at other majors that are in the technological field that can be medical or engineering.

**Does final GPA differ across majors?** Read the exceptions, not just the boxes.

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots(figsize (9,5))
#   sns.boxplot(data=students, x="final_gpa", y="major", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("Final GPA by major - boxes agree, outliers don't")
#   ax.set_xlabel("Final GPA")
#   ax.set_ylabel("Major")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots(figsize=(9,5))
sns.boxplot(data=students, x="final_gpa", y="major", color=NAVY, ax=ax)
ax.set_title("Final GPA by major - boxes agree, outliers don't")
ax.set_xlabel("Final GPA")
ax.set_ylabel("Major")
plt.show()
```

### Q4 · Two quantitative variables
**Is there a relationship between how far students commute and their final GPA?** Remember: *association is not causation.*
→ *scatterplot with a trend line*. Columns: `commute_miles` and `final_gpa` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#   Rust = "#C0562F"
#   fig, ax = plt.subplots()
#   sns.regplot(data=students, x="commute_miles", y="final_gpa", color=NAVY, ax=ax)
#                 scatter_kws={"s":15, "color":NAVY, "alpha":0.5},
#                 line_kws={"color":RUST}, ax=ax)
#   ax.set_title("commute_miles vs. final_gpa")
#   ax.set_xlabel("Commute Miles")
#   ax.set_ylabel("Final GPA")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots()
sns.regplot(data=students, x="commute_miles", y="final_gpa",
            scatter_kws={"s":15, "color":NAVY, "alpha":0.5},
            line_kws={"color":RUST},
            ax=ax)
ax.set_title("Commute Miles vs. Final GPA")
ax.set_xlabel("Commute Miles")
ax.set_ylabel("Final GPA")
plt.show()
```

### Q5 · Over time
**How does student engagement change across the semester?**
→ *line chart* of average weekly activity. Use the `weekly` table: `week` and `minutes_active` (you'll need to average by week first — see the Guided notebook).

```python
weekly_avg = weekly.groupby("week")["minutes_active"].mean().reset_index()
fig, ax = plt.subplots()
sns.lineplot(data=weekly_avg, x="week", y="minutes_active", color=NAVY, marker="o", ax=ax)
ax.set_title("Average Weekly Activity")
ax.set_xlabel("Week")
ax.set_ylabel("Avg. Minutes Active")
plt.show()
```

**STAR story — Q5** *(write in your own words)*

- **Scope —** Looking at the line graph it consists of weekly activity and average minutes active for students.
- **Track —** The question is stating how student engagement change throughtout the semester at Northgate.
- **Articulate —** The average weekly activities are higher in the 200 minute section. Unlike between week 8 and 10 it drops heavily under 100 minutes then rises up again going into weeks 11-16.
- **Respond —** What was odd and unexpected was the averages being so consistent with the amount of people working above average hours in the earlier chart.
